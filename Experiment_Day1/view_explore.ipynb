# View and Explore TSV Files
This notebook displays the top 30 rows of all `.tsv` files found in the `train` and `test` folders.

```python
import pandas as pd
import glob
import os

# Path to the dataset directory
dataset_dir = '../student_resource/dataset'

# Find all TSV files in train and test folders
train_files = glob.glob(os.path.join(dataset_dir, 'train', '*.tsv'))
test_files = glob.glob(os.path.join(dataset_dir, 'test', '*.tsv'))
tsv_files = train_files + test_files

for file_path in tsv_files:
    print(f"\n{'='*80}")
    print(f"File: {file_path}")
    print(f"{'='*80}")
    try:
        # Read and display top 30 rows
        df = pd.read_csv(file_path, sep='\t', nrows=50)
        display(df)
    except Exception as e:
        print(f"Error reading {file_path}: {e}")
```

## Dataset Sizes
Counting the number of rows for each TSV file in the dataset.

```python
import glob
import os

dataset_dir = '../student_resource/dataset'
tsv_files = glob.glob(os.path.join(dataset_dir, '**', '*.tsv'), recursive=True)

for f in sorted(tsv_files):
    with open(f, 'rb') as file:
        count = sum(1 for _ in file) - 1 # Subtract 1 for header
        print(f"{os.path.basename(f)}: {count:,} rows")
```
